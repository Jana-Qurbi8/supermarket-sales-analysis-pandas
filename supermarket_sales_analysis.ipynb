# Descriptive Statistics

In this section, we calculate basic statistical measures for the Sales column.

```python
print("Average Sales:", df['Sales'].mean())
print("Median Sales:", df['Sales'].median())
print("Maximum Sales:", df['Sales'].max())
```

# Business Questions

## Q1. Which branch had the highest total sales?

```python
df.groupby('Branch')['Sales'].sum().sort_values(ascending=False)
```

# Q2. Which city generated the highest total sales?

```python
df.groupby('City')['Sales'].sum().sort_values(ascending=False)
```

# Q3. Which product line generated the highest total sales?

```python
df.groupby('Product line')['Sales'].sum().sort_values(ascending=False)
```

# Q4. Which payment method was used the most?


```python
df.groupby('Payment')['Payment'].count().sort_values(ascending=False)
```

# Q5. Which branch has the highest average customer rating?

```python
df.groupby('Branch')['Rating'].mean().sort_values(ascending=False)
```

# Q6. Which product line has the highest average customer rating?

```python
df.groupby('Product line')['Rating'].mean().sort_values(ascending=False)
```
